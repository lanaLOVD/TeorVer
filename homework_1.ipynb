### Задача 1.8

> "В конкурсе по 5 номинациям участвуют 10 кинофильмов.Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?"



```python
setFilms = ['F1','F2','F3','F4','F5','F6','F7','F8','F9','F10']
k = 5

#разные подарки

nominationA = list(product(setFilms, repeat=k))
print('Всего возможных вариантов (разные призы): {}.'.format(len(nominationA)))

#одинаковые подарки

nominationB = list(combinations_with_replacement(setFilms, r=k))
print('Всего возможных вариантов (одинаковые призы): {}.'.format(len(nominationB)))
```

### Пример 1.10. 
> Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикладывает
одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность
того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

```python
letters = ['Т', 'Е', 'И', 'Я', 'Р', 'О']

# Пункт а
all_3 = list(permutations(letters, 3))
count_a = 1
print(f"а) {count_a}/{len(all_3)} = {count_a/len(all_3)}")

# Пункт б
all_6 = list(permutations(letters))
count_b = 1  
print(f"б) {count_b}/{len(all_6)} = {count_b/len(all_6)}") # 210
```

### Пример 1.11. 
> Используя условие примера 1.10, найти вероятность
того, что получится слово «АНАНАС», если на отдельных
карточках написаны три буквы А, две буквы Н и одна буква С.

```python
letters = ['А', 'А', 'А', 'Н', 'Н', 'С']

unique = set(permutations(letters))

total = len(unique)
print(f"Общее число исходов: {total}")

target = ('А', 'Н', 'А', 'Н', 'А', 'С')
count = 1 if target in unique else 0
print(f"Благоприятных: {count}")

print(f"Вероятность: {count}/{total} = {count/total}")
```

### Пример 1.16. 
> За круглым столом рассаживаются 5 мужчин и
5 женщин. Найти вероятность того, что: а) никакие два лица одного
пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

```python
total = math.factorial(9)  

# Пункт а
favorable_a = math.factorial(4) * math.factorial(5)
a = favorable_a/total
print(f"а) {a:4f}")  

# Пункт б
favorable_b = math.factorial(4) * (2**5)  
b = favorable_b/total
print(f"б) {b:4f}") 
```

### Пример 1.17. 
> В купейный вагон (9 купе по 4 места) семи пассажирам
продано 7 билетов. Найти вероятности того, что пассажиры
попали: а) в два купе; б) в семь купе; в) в три купе.

```python
def A(n, k):
    if k > n: return 0
    return math.factorial(n) // math.factorial(n - k)

total = A(36, 7)

# а) в 2 купе
k = 2
M_a = math.comb(9, k) * A(4*k, 7)
a = M_a/total
print(f"а) {a:4f}")

# б) в 7 купе
k = 7
M_b = math.comb(9, k) * math.factorial(7) * (4**7)
b=M_b/total
print(f"б) {b:4f}")

# в) в 3 купе
k = 3
M_v = math.comb(9, k) * (A(4*k, 7) - math.comb(k, 1) * A(4*(k-1), 7))
v=M_v/total
print(f"в) {v:4f}")
```
