> Пусть из первой урны переложили k белых шаров (k = 0, 1, 2, 3).
Вероятность этого считаем через сочетания. После перекладывания во второй урне 10 шаров, из них белых 6 + k.
Дальше формула полной вероятности. 

```python
from math import comb

total = comb(20, 3) 
answer = 0

for k in range(4):
    p_k = comb(15, k) * comb(5, 3 - k) / total  
    p_white = (6 + k) / 10                 
    answer += p_k * p_white

print("Классический ответ ", answer)
```

## Моделирование

```python
import random

def one_try():
    urn1 = ["w"] * 15 + ["b"] * 5
    urn2 = ["w"] * 6 + ["b"] * 1

    moved = random.sample(urn1, 3)  
    urn2 += moved                

    ball = random.choice(urn2)    
    return ball == "w"

n = 1000000
count = 0
for i in range(n):
    if one_try():
        count += 1

print("Моделирование ", count / n)
```
